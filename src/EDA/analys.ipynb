```python
df = pd.read_csv('../../data/metadata.csv')
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df = df[(df != '').all(axis=1)]
```

```python
len(np.unique(df['tag']))
```

```python
gp = pd.read_csv('../../data/graph.csv')
```

```python
df.to_csv('../../data/metadata.csv', index=False)
```

```python
gp.target.value_counts()
```

```python
# Получить уникальные значения из колонки 'tag'
unique_subcategories = df['tag'].unique()

# Создать отображения: категория -> индекс
subcat2idx = {subcategory: idx for idx, subcategory in enumerate(unique_subcategories)}
idx2subcat = {idx: subcategory for idx, subcategory in enumerate(unique_subcategories)}

# Преобразовать колонку 'tag' в индексы
df_indices = df['tag'].map(subcat2idx)

# Найти значения, которые есть в column1 (df['tag']), но отсутствуют в column2 (gp['target'])
missing_in_column2 = set(df_indices) - set(gp['target'])
print("Значения из df, которых нет в gp:", missing_in_column2)

# Найти значения, которые есть в column2 (gp['target']), но отсутствуют в column1 (df['tag'])
missing_in_column1 = set(gp['target']) - set(df_indices)
print("Значения из gp, которых нет в df:", missing_in_column1)
```

```python
miser = []
for idx, mis in enumerate(missing_in_column2):
    miser.append(idx2subcat[mis])
print(miser)
```

```python
# Удаление строк, где 'tag' соответствует значениям из missing_in_gp
df_filtered = df[~df['tag'].isin(miser)]
```

```python
df_filtered.tag.value_counts()
```

```python
df_filtered.to_csv('../../data/metadata.csv', index=False)
```
